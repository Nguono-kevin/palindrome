```python
rows = 1  # Initialize rows

while rows < 5:  # Outer loop for rows
    cols = 1  # Reset cols for each row

    while cols <= rows:  # Inner loop for columns
        print("*", end="")  # Print star without a newline
        cols += 1  # Increment column

    print("")  # Move to the next line after printing stars
    rows += 1  # Increment row
```

```python
rows=5
while rows > 0:
     cols =0
     while cols< rows:
         print("*", end="")
         cols +=1
     print("")
     rows -=1
```

```python
rows = 5  # Set the number of rows

while rows > 0:  # Outer loop for rows
    spaces = 5 - rows  # Calculate spaces before stars
    cols = rows  # Number of stars in each row

    print(" " * spaces + "*" * cols)  # Print spaces and stars

    rows -= 1  # Decrease row count
```

```python
rows = 4  # Total number of rows

for i in range(1, rows + 1):  # Loop through 1 to 4
    spaces = rows - i  # Number of spaces before stars
    stars = i  # Number of stars in each row
    print(" " * spaces + "*" * stars)  # Print spaces first, then stars
```
